```python

import re
from collections import Counter, defaultdict
from difflib import get_close_matches

# Sample training text
corpus = """
machine learning is a part of artificial intelligence
artificial intelligence is changing the world
python is a popular programming language
machine learning helps computers learn from data
natural language processing helps computers understand language
deep learning is used in artificial intelligence
python programming is useful for machine learning
data science uses python for data analysis
"""

# Tokenization
words = re.findall(r'\b[a-z]+\b', corpus.lower())

# Build vocabulary
vocabulary = set(words)

# Build Bigram Model
bigram_model = defaultdict(Counter)

for i in range(len(words) - 1):
    bigram_model[words[i]][words[i + 1]] += 1

# Spell correction
def autocorrect(word):
    word = word.lower()

    if word in vocabulary:
        return word

    suggestions = get_close_matches(
        word, vocabulary, n=1, cutoff=0.65
    )

    if suggestions:
        return suggestions[0]

    return word

# Next word prediction
def predict_next_word(sentence):
    tokens = re.findall(r'\b[a-z]+\b', sentence.lower())

    if not tokens:
        return "Enter a sentence first."

    last_word = tokens[-1]

    if last_word in bigram_model:
        predictions = bigram_model[last_word].most_common(3)
        return [word for word, count in predictions]

    return ["No prediction available"]

# Complete system
def smart_keyboard(sentence):
    corrected_words = [
        autocorrect(word)
        for word in re.findall(r'\b[a-z]+\b', sentence.lower())
    ]

    corrected_sentence = " ".join(corrected_words)

    predictions = predict_next_word(corrected_sentence)

    print("\nOriginal Text:", sentence)
    print("Corrected Text:", corrected_sentence)
    print("Next Word Suggestions:", predictions)

# Test the system
smart_keyboard("machin learning is")
smart_keyboard("pythn is a popular")
smart_keyboard("artifical intelligence is")
```
