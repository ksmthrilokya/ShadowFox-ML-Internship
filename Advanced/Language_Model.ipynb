
# Text Generation and Language Model Performance Analysis using DistilGPT-2

## 1. Introduction

A Language Model (LM) is an Artificial Intelligence model that understands patterns in human language and predicts the next word or token based on the given context.

In this project, we use DistilGPT-2, a lightweight transformer-based language model, to generate text from user-provided prompts.

## Tools and Technologies

- DistilGPT-2


```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

print("DistilGPT-2 loaded successfully!")
```

```python
prompt = "Artificial Intelligence is transforming the world by"

result = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```

```python
prompt = "Artificial Intelligence is transforming the world by"

for temp in [0.2, 0.7, 1.2]:
    result = generator(
        prompt,
        max_new_tokens=40,
        temperature=temp,
        do_sample=True,
        num_return_sequences=1
    )

    print("\nTemperature:", temp)
    print(result[0]["generated_text"])
    print("-" * 60)
```

### Experiment 2: Effect of Temperature on Text Generation

**Objective:** To study how temperature affects the creativity and predictability of generated text.

**Input Prompt:**  
"Artificial Intelligence is transforming the world by"

**Temperature values tested:** 0.2, 0.7, 1.2.

**Observations:**
1. At temperature 0.2, the model generated predictable but repetitive text.
2. At temperature 0.7, the output was relatively balanced and readable.
3. At temperature 1.2, the output was more diverse but less focused.

**Conclusion:** Temperature controls the randomness of text generation. Lower values produce more predictable outputs, while higher values increase variation and may reduce coherence.

```python
prompts = [
    "Machine Learning is useful because",
    "The future of robotics",
    "Education in the digital age"
]

for prompt in prompts:
    result = generator(
        prompt,
        max_new_tokens=35,
        do_sample=True,
        temperature=0.7
    )

    print("\nPrompt:", prompt)
    print("Generated:", result[0]["generated_text"])
    print("-" * 50)
```

### Experiment 3: Testing Different Input Prompts

**Objective:** To evaluate how DistilGPT-2 generates text for different input topics.

**Prompts Tested:**
1. Machine Learning is useful because
2. The future of robotics
3. Education in the digital age

**Observations:**
- For the Machine Learning prompt, the model generated related content but repeated some phrases.
- For the Robotics prompt, it generated a short statement about the future of robotics.
- For the Education prompt, it generated a response related to digital education but lacked clarity.

**Conclusion:** DistilGPT-2 can generate text for different topics based on input prompts. However, its responses may contain repetition, irrelevant statements, or grammatical inconsistencies.

**Analysis:** Lower temperature produces more predictable text, whereas higher temperature produces more varied and creative text. However, higher temperature may reduce the coherence of the generated content.

## Strengths and Limitations of DistilGPT-2

### Strengths
1. Generates human-like text based on input prompts.
2. Supports different text-generation tasks.
3. Requires relatively fewer computational resources than larger GPT models.
4. Produces outputs quickly for simple prompts.
5. Can be used for experimenting with Natural Language Processing.

### Limitations
1. May generate repetitive sentences.
2. Can produce grammatically incorrect or unclear text.
3. May generate incorrect or misleading information.
4. Does not always maintain proper context.
5. Its knowledge may be outdated or incomplete.

### Overall Analysis
DistilGPT-2 is useful for understanding language generation and experimenting with NLP. However, its generated content should be reviewed because it may contain errors, repetition, and misleading information.

## Ethical Considerations

2. **Bias:** Generated text may reflect biases present in its training data.
3. **Privacy:** Users should avoid entering sensitive personal information into language models.
4. **Responsible Usage:** Generated content should be checked before using or publishing it.

## Model Parameters and Their Functions

| Parameter | Description |
|---|---|
| max_new_tokens | Controls the maximum number of new tokens generated. |
| temperature | Controls the randomness of generated text. |
| do_sample | Enables sampling-based text generation when set to True. |
| num_return_sequences | Specifies the number of generated outputs. |

**Analysis:** These parameters help control the length, creativity, and number of generated text outputs. Changing their values can produce different results from the same input prompt.
